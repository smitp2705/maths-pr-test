## Part A – Theory (Short Questions)

---

### Q1. Explain Mean, Median, and Mode with a real-life example.

**Mean** is the arithmetic average of all values — sum divided by count.  
**Median** is the middle value when data is sorted in order.  
**Mode** is the value that appears most frequently in the dataset.

**Real-life example:** Suppose 5 students score: 45, 60, 60, 75, 90.
- Mean = (45+60+60+75+90)/5 = **66**
- Median = middle value = **60**
- Mode = most repeated value = **60**

In salary analysis, median is preferred over mean because one extremely high salary can skew the mean, making the median a more representative "typical" value.

---

### Q2. What is the difference between Variance and Standard Deviation?

**Variance** measures the average squared deviation of each data point from the mean. It gives a sense of how spread out the data is, but in squared units, which can be hard to interpret directly.

**Standard Deviation (SD)** is simply the square root of variance, bringing the measure back to the original units of the data — making it more intuitive.

**Example:** If Math_Scores have a variance of 225, the standard deviation is √225 = 15, meaning scores typically deviate 15 points from the mean.

Variance is useful in mathematical derivations (e.g., ANOVA), while SD is preferred for reporting and communication.

---

### Q3. Define Normal Distribution and give one practical use case.

A **Normal Distribution** is a symmetric, bell-shaped probability distribution where data is evenly spread around the mean. It is defined by two parameters: mean (μ) and standard deviation (σ). The empirical rule states that ~68% of data falls within ±1σ, ~95% within ±2σ, and ~99.7% within ±3σ.

**Practical use case:** Student exam scores often follow a normal distribution. If the mean score is 65 and σ = 10, a teacher can predict that roughly 95% of students will score between 45 and 85. This helps set grading thresholds fairly and identify outliers (very high or very low performers).

---

### Q4. Explain Skewness and Kurtosis in simple words.

**Skewness** measures the asymmetry of a distribution.
- **Positive skew (right skew):** Tail extends to the right; most values are low, but a few are very high (e.g., income distribution).
- **Negative skew (left skew):** Tail extends to the left; most values are high (e.g., age at retirement).
- Zero skewness = perfectly symmetric (normal distribution).

**Kurtosis** measures the "peakedness" or heaviness of the tails of a distribution.
- **High kurtosis (leptokurtic):** Very sharp peak, heavy tails — extreme values are more likely.
- **Low kurtosis (platykurtic):** Flat peak, thin tails — values are spread more evenly.
- Normal distribution has a kurtosis of 3 (excess kurtosis = 0).

---

### Q5. What is Probability? Differentiate between Empirical vs Theoretical Probability.

**Probability** is a numerical measure (between 0 and 1) of how likely an event is to occur. P=0 means impossible, P=1 means certain.

| | Theoretical Probability | Empirical Probability |
|---|---|---|
| **Definition** | Based on known outcomes assuming equal likelihood | Based on observed data from experiments |
| **Formula** | Favorable outcomes / Total outcomes | Occurrences of event / Total observations |
| **Example** | P(Heads) = 1/2 on a fair coin | Flip a coin 100 times; get 48 heads → P(H) = 0.48 |
| **When used** | When outcomes are equally likely & known | When real-world data is available |

In our dataset, the empirical probability of passing = number of students with Pass_Fail=1 / total students.

---

### Q6. Explain Independent vs Dependent Events with one example each.

**Independent Events:** Two events are independent if the occurrence of one does NOT affect the probability of the other.
- **Example:** Rolling a die and flipping a coin. Getting a 6 on the die does not change the chance of getting heads on the coin. P(A∩B) = P(A) × P(B).

**Dependent Events:** Two events are dependent if the occurrence of one DOES affect the probability of the other.
- **Example:** Drawing cards from a deck without replacement. If you draw an Ace first, the probability of drawing another Ace changes because the deck now has one fewer Ace. P(A∩B) = P(A) × P(B|A).

In education, studying more hours and passing are **dependent** events — studying increases the probability of passing.

---

### Q7. What is the intuition of Bayes Theorem in daily life?

**Bayes' Theorem** helps us update our belief about an event based on new evidence. The formula is:

> **P(A|B) = [P(B|A) × P(A)] / P(B)**

**Daily life intuition (Medical Test Example):** Suppose a disease affects 1% of the population. A test is 95% accurate. If you test positive, what is the actual probability you have the disease?

Bayes' Theorem tells us to combine: (1) our prior belief (1% base rate), (2) the likelihood of the test result given disease (95%), and (3) the overall probability of testing positive. The result is often surprising — even with a positive test, the probability might be much lower than 95% because the disease is rare.

In short, Bayes helps us reason from evidence to cause — it's the engine behind spam filters, medical diagnosis, and recommendation systems.

---

### Q8. Explain Eigenvalue and Eigenvector in simple terms.

Imagine a transformation (matrix) that stretches, rotates, or scales data. An **eigenvector** is a special direction that does NOT change direction when that transformation is applied — it only gets scaled. The **eigenvalue** is the amount by which it gets scaled.

**Simple analogy:** Think of a rubber band stretched along one axis. The direction of stretch is the eigenvector; how much it stretches is the eigenvalue.

**Practical use:** In **Principal Component Analysis (PCA)**, eigenvectors of the covariance matrix define the new axes (principal components) along which data varies the most. Eigenvalues tell us how much variance each component explains — helping reduce dimensionality while retaining the most information.

## Part B – Practical (Python Programming)

### Setup: Import Libraries & Load Data

```python
# ── Imports ──────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import norm
import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

# ── Load Dataset ──────────────────────────────────────────────────────────
df = pd.read_csv('students_scores.csv')
print('Shape:', df.shape)
df.head()
```

```python
# Basic info & null check
print(df.info())
print('\nNull values:\n', df.isnull().sum())
print('\nDescriptive Stats:')
df.describe().round(2)
```

---
## Step 1: Measures of Central Tendency & Dispersion

```python
# ── 1a. Mean, Median, Mode of Math_Score ─────────────────────────────────
math_mean   = df['Math_Score'].mean()
math_median = df['Math_Score'].median()
math_mode   = df['Math_Score'].mode()[0]

print('===== Math_Score: Central Tendency =====')
print(f'  Mean   : {math_mean:.2f}')
print(f'  Median : {math_median:.2f}')
print(f'  Mode   : {math_mode:.2f}')

# Visualise
fig, ax = plt.subplots()
ax.hist(df['Math_Score'], bins=30, color='steelblue', edgecolor='white', alpha=0.85)
ax.axvline(math_mean,   color='red',    linestyle='--', lw=2, label=f'Mean = {math_mean:.1f}')
ax.axvline(math_median, color='green',  linestyle='-.',  lw=2, label=f'Median = {math_median:.1f}')
ax.axvline(math_mode,   color='orange', linestyle=':',   lw=2, label=f'Mode = {math_mode:.1f}')
ax.set_title('Math_Score Distribution with Central Tendency Lines')
ax.set_xlabel('Math Score')
ax.set_ylabel('Frequency')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ── 1b. Range, Variance, Standard Deviation of Science_Score ─────────────
sci_range  = df['Science_Score'].max() - df['Science_Score'].min()
sci_var    = df['Science_Score'].var()
sci_std    = df['Science_Score'].std()

print('===== Science_Score: Dispersion Measures =====')
print(f'  Range              : {sci_range:.2f}')
print(f'  Variance           : {sci_var:.2f}')
print(f'  Standard Deviation : {sci_std:.2f}')

# Box plot to visualise spread
fig, ax = plt.subplots()
ax.boxplot(df['Science_Score'].dropna(), vert=False, patch_artist=True,
           boxprops=dict(facecolor='lightcoral', color='darkred'),
           medianprops=dict(color='darkred', lw=2))
ax.set_title('Science_Score Box Plot (Spread Visualisation)')
ax.set_xlabel('Science Score')
plt.tight_layout()
plt.show()
```

---
## Step 2: Probability Basics

```python
# ── 2a. Probability of Passing (Pass_Fail = 1) ────────────────────────────
total_students = len(df)
passing = df['Pass_Fail'].sum()
p_pass = passing / total_students

print('===== Probability of Passing =====')
print(f'  Total Students  : {total_students}')
print(f'  Passing Students: {int(passing)}')
print(f'  P(Pass)         : {p_pass:.4f}  ({p_pass*100:.2f}%)')

# Pie chart
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie([passing, total_students - passing],
       labels=['Pass', 'Fail'],
       autopct='%1.1f%%',
       colors=['#4CAF50', '#F44336'],
       startangle=90,
       explode=(0.05, 0))
ax.set_title('Pass / Fail Distribution')
plt.tight_layout()
plt.show()
```

```python
# ── 2b. Contingency Table: Pass_Fail vs Hours_Studied > 5 ─────────────────
df['High_Study'] = (df['Hours_Studied'] > 5).astype(int)
contingency = pd.crosstab(df['Pass_Fail'], df['High_Study'],
                           rownames=['Pass_Fail (0=Fail, 1=Pass)'],
                           colnames=['Hours_Studied > 5 (0=No, 1=Yes)'],
                           margins=True, margins_name='Total')
print('===== Contingency Table =====')
print(contingency)

# Heatmap
fig, ax = plt.subplots(figsize=(6, 4))
sns.heatmap(pd.crosstab(df['Pass_Fail'], df['High_Study']),
            annot=True, fmt='d', cmap='Blues',
            xticklabels=['Low Study (≤5h)', 'High Study (>5h)'],
            yticklabels=['Fail', 'Pass'], ax=ax)
ax.set_title('Contingency Table Heatmap: Pass/Fail vs Study Hours')
plt.tight_layout()
plt.show()
```

```python
# ── 2c. Conditional Probability: P(Pass | Hours_Studied > 5) ─────────────
high_study    = df[df['Hours_Studied'] > 5]
pass_given_hs = high_study['Pass_Fail'].sum() / len(high_study)

low_study = df[df['Hours_Studied'] <= 5]
pass_given_ls = low_study['Pass_Fail'].sum() / len(low_study)

print('===== Conditional Probabilities =====')
print(f'  P(Pass | Hours_Studied > 5)  : {pass_given_hs:.4f}  ({pass_given_hs*100:.1f}%)')
print(f'  P(Pass | Hours_Studied ≤ 5)  : {pass_given_ls:.4f}  ({pass_given_ls*100:.1f}%)')
print(f'  Lift (High vs Low study)     : {pass_given_hs/pass_given_ls:.2f}x')

# Bar chart
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(['Hours > 5 (High Study)', 'Hours ≤ 5 (Low Study)'],
       [pass_given_hs, pass_given_ls],
       color=['#4CAF50', '#F44336'], edgecolor='white', width=0.5)
ax.set_ylabel('P(Pass)')
ax.set_title('Conditional Probability of Passing by Study Hours')
ax.set_ylim(0, 1)
for i, v in enumerate([pass_given_hs, pass_given_ls]):
    ax.text(i, v + 0.02, f'{v:.1%}', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

## Step 3: Distribution & Visualization

```python
# ── 3a. Histogram + Normal Curve for Math_Score ────────────────────────────
mu, sigma = df['Math_Score'].mean(), df['Math_Score'].std()

fig, ax = plt.subplots()
# Histogram (density=True so normal curve overlays correctly)
ax.hist(df['Math_Score'], bins=30, density=True,
        color='steelblue', edgecolor='white', alpha=0.7, label='Histogram')
# Normal curve
x = np.linspace(df['Math_Score'].min(), df['Math_Score'].max(), 300)
ax.plot(x, norm.pdf(x, mu, sigma), color='crimson', lw=2.5,
        label=f'Normal Curve\nμ={mu:.1f}, σ={sigma:.1f}')
ax.set_title('Math_Score: Histogram + Normal Curve')
ax.set_xlabel('Math Score')
ax.set_ylabel('Density')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ── 3b. Skewness & Kurtosis for Science_Score ─────────────────────────────
sci_skew = df['Science_Score'].skew()
sci_kurt = df['Science_Score'].kurt()   # excess kurtosis (normal = 0)

print('===== Science_Score: Shape Statistics =====')
print(f'  Skewness         : {sci_skew:.4f}')
print(f'  Kurtosis (excess): {sci_kurt:.4f}')

if sci_skew > 0.5:
    skew_interp = 'Positively skewed (right tail)'
elif sci_skew < -0.5:
    skew_interp = 'Negatively skewed (left tail)'
else:
    skew_interp = 'Approximately symmetric'

kurt_interp = 'Leptokurtic (heavy tails)' if sci_kurt > 0 else 'Platykurtic (light tails)'
print(f'  Interpretation   : {skew_interp} | {kurt_interp}')

# Visualise
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['Science_Score'], bins=30, color='mediumpurple', edgecolor='white', alpha=0.8)
axes[0].set_title(f'Science_Score Distribution\nSkew={sci_skew:.2f}, Kurt={sci_kurt:.2f}')
axes[0].set_xlabel('Science Score')
axes[0].set_ylabel('Frequency')

axes[1].boxplot(df['Science_Score'].dropna(), patch_artist=True,
                boxprops=dict(facecolor='mediumpurple', alpha=0.6),
                medianprops=dict(color='black', lw=2))
axes[1].set_title('Science_Score Box Plot')
axes[1].set_ylabel('Score')

plt.tight_layout()
plt.show()
```

```python
# ── 3c. Q-Q Plot for English_Score ────────────────────────────────────────
fig, ax = plt.subplots(figsize=(6, 6))
(osm, osr), (slope, intercept, r) = stats.probplot(df['English_Score'].dropna(), dist='norm')
ax.scatter(osm, osr, color='teal', alpha=0.4, s=10, label='Data')
ax.plot(osm, slope * np.array(osm) + intercept, color='red', lw=2, label='Reference Line')
ax.set_title(f'Q-Q Plot: English_Score\n(R = {r:.4f})')
ax.set_xlabel('Theoretical Quantiles')
ax.set_ylabel('Sample Quantiles')
ax.legend()
plt.tight_layout()
plt.show()

print('Q-Q Plot Interpretation:')
if abs(r) > 0.99:
    print('  Points closely follow the reference line → English_Score is approximately normally distributed.')
else:
    print('  Points deviate from the reference line → English_Score shows departures from normality.')
```

## Step 4: Linear Algebra Mini Task

```python
# ── 4. First 5 Students as Vectors ────────────────────────────────────────
math_vec    = df['Math_Score'].values[:5].astype(float)
science_vec = df['Science_Score'].values[:5].astype(float)

print('===== Vectors (First 5 Students) =====')
print(f'  Math_Score    vector: {math_vec}')
print(f'  Science_Score vector: {science_vec}')

# ── Dot Product ────────────────────────────────────────────────────────────
dot_product = np.dot(math_vec, science_vec)
print(f'\n  Dot Product (Math · Science): {dot_product:.2f}')

# ── Norm 1 and Norm 2 of Math_Score Vector ─────────────────────────────────
norm1 = np.linalg.norm(math_vec, ord=1)   # L1 norm (Manhattan)
norm2 = np.linalg.norm(math_vec, ord=2)   # L2 norm (Euclidean)
print(f'\n  Math_Score Vector Norms:')
print(f'    Norm 1 (L1 / Manhattan) : {norm1:.2f}')
print(f'    Norm 2 (L2 / Euclidean) : {norm2:.2f}')

# ── Angle Between Vectors ──────────────────────────────────────────────────
cos_theta = dot_product / (norm2 * np.linalg.norm(science_vec, ord=2))
cos_theta = np.clip(cos_theta, -1, 1)       # numerical safety
angle_rad = np.arccos(cos_theta)
angle_deg = np.degrees(angle_rad)

print(f'\n  Angle Between Vectors:')
print(f'    cos(θ)   : {cos_theta:.6f}')
print(f'    θ (rad)  : {angle_rad:.4f}')
print(f'    θ (deg)  : {angle_deg:.2f}°')

if angle_deg < 30:
    print('    → Very similar directions: Math & Science scores are highly correlated for first 5 students.')
elif angle_deg < 60:
    print('    → Moderately similar directions.')
else:
    print('    → Dissimilar directions: Math & Science scores diverge for first 5 students.')
```

```python
# ── Visual Summary of Vector Operations ────────────────────────────────────
students_idx = [f'S{i+1}' for i in range(5)]

x = np.arange(5)
width = 0.35

fig, ax = plt.subplots(figsize=(9, 4))
bars1 = ax.bar(x - width/2, math_vec,    width, label='Math_Score',    color='steelblue')
bars2 = ax.bar(x + width/2, science_vec, width, label='Science_Score', color='mediumpurple')

ax.set_xticks(x)
ax.set_xticklabels(students_idx)
ax.set_xlabel('Student')
ax.set_ylabel('Score')
ax.set_title(f'First 5 Students — Math & Science Vectors\n'
             f'Dot Product={dot_product:.0f} | Angle={angle_deg:.1f}° | ||Math||₂={norm2:.1f}')
ax.legend()
plt.tight_layout()
plt.show()
```
